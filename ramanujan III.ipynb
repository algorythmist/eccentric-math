# Redefining Convergence


We have now seen a couple of proofs that are often sited but are nonetheless wrong. Now we get to the correct proof. Since we have already shown that the sum of the sequence $1+2+3+… $diverges, the only way to get around that is to redefine by what we mean by the convergence of a sum. We will examine three different generalizations of the definition of the sum of an infinite sequence, each one more liberal.

### Cesaro Summmation

As before, let $a_n$ be a sequence and let the sequence of partial sums be

$$
s_n := \sum_{k=0}^n a_k
$$

The sequence $a_n$ is called Cesàro summable, with Cesàro sum 
$A ∈ R {\displaystyle \mathbb {R} } \mathbb {R}$ , if, as $n$ tends to infinity, 
the arithmetic mean of its first $n$ partial sums $s_1, s_2, ..., s_n$ tends to $A$:

$$
\lim_{n → ∞} \frac{1}{n}\sum_{k = 1}^n s_k = A
$$

#### Example: Grandi's series is Cesàro summable

Recall from Part I the sequence $a_n = 1,-1,1,-1,\ldots$, with $A = 1-1+1-1+\ldots$

The partial sums are

$$
s_n = 1,0,1,0,1,\ldots
$$

which oscillate forever and never settle down, so the sequence $s_n$ does not converge. In the ordinary sense, $A$ is not summable.

Now form the Cesàro means, i.e. the running average of the $s_n$:

$$
\frac{1}{n}\sum_{k=1}^n s_k
$$

For $n=1,2,3,4,5,6,\ldots$ this average works out to

$$
1, \frac{1}{2}, \frac{2}{3}, \frac{1}{2}, \frac{3}{5}, \frac{1}{2}, \ldots
$$

Unlike $s_n$ itself, this sequence of averages does converge, and it converges to $\frac{1}{2}$. So while $A$ has no ordinary sum, it is Cesàro summable with Cesàro sum $\frac{1}{2}$ — matching the (fallacious) value we obtained for $A$ in Part I, but now on solid footing.</cell id="6c17ec94">


### Abel Summation

Cesàro summation is not the only way to generalize convergence, and it turns out not to be liberal enough for our purposes — the sequence $1+2+3+\ldots$ is not Cesàro summable either (its partial sums $s_n$ grow without bound, and so do their running averages).

Abel summation offers a more powerful generalization. Given a sequence $a_n$, form the power series

$$
f(x) := \sum_{n=0}^{\infty} a_n x^n
$$

For $0 < x < 1$, the factor $x^n$ shrinks fast enough that this power series may converge even when $\sum a_n$ itself does not. If $f(x)$ converges for all $0<x<1$ and the limit

$$
A := \lim_{x \to 1^-} f(x)
$$

exists, we say that $a_n$ is Abel summable with Abel sum $A$.</cell id="85753f3b">


#### Example: $1-2+3-4+\ldots$ is Abel summable

Recall from Part I the sequence $B = 1-2+3-4+\ldots$, i.e. $a_n = (-1)^n(n+1)$ for $n=0,1,2,\ldots$

Form the power series

$$
f(x) = \sum_{n=0}^{\infty} (-1)^n (n+1) x^n = 1 - 2x + 3x^2 - 4x^3 + \ldots
$$

This is a standard power series with closed form

$$
f(x) = \frac{1}{(1+x)^2}, \qquad 0 \le x < 1
$$

Taking the limit as $x \to 1^-$,

$$
\lim_{x \to 1^-} f(x) = \frac{1}{(1+1)^2} = \frac{1}{4}
$$

So $B$ is Abel summable with Abel sum $\frac{1}{4}$ — again matching the value we obtained fallaciously in Part I, but this time legitimately. Note that $B$ is *not* Cesàro summable: its partial sums $s_n = 1,-1,2,-2,3,-3,\ldots$ grow without bound, so their running averages do not converge either. Abel summation genuinely extends Cesàro summation to a larger class of series.</cell id="85753f3b">


### A Bridge to Bernoulli Numbers

Before moving on to Ramanujan summation in Part IV, it's worth meeting the *Bernoulli numbers* $B_0, B_1, B_2, \ldots$ somewhere safe first — in an ordinary, finite, perfectly convergent sum, rather than in the middle of a divergent one.

They show up in **Faulhaber's formula**, the closed-form expression for sums of integer powers:

$$
\sum_{k=1}^{n} k^p = \frac{1}{p+1}\sum_{j=0}^{p} \binom{p+1}{j} B_j\, n^{p+1-j}
$$

(using the convention $B_1 = +\frac{1}{2}$, appropriate when the sum starts at $k=1$; some sources use $B_1=-\frac12$ for sums starting at $k=0$ instead — this sign ambiguity in $B_1$ is a real thing to watch for, though it won't matter for what follows, since $B_2$ onward is the same either way).

#### Example: $p=1$, the sum of the first $n$ integers

We can derive this case elementarily, without invoking Bernoulli numbers at all, using a telescoping trick. Since $(k+1)^2 - k^2 = 2k+1$, summing both sides from $k=1$ to $n$ telescopes on the left:

$$
(n+1)^2 - 1 = 2\sum_{k=1}^{n} k + n
$$

Solving for the sum,

$$
\sum_{k=1}^{n} k = \frac{n^2+n}{2} = \frac{n(n+1)}{2}
$$

This is exactly the $p=1$ case of Faulhaber's formula: $\frac{1}{2}\big(B_0 n^2 + 2B_1 n\big) = \frac{1}{2}\big(n^2 + n\big)$, using $B_0=1$, $B_1=\frac12$.

#### Example: $p=2$, the sum of the first $n$ squares — where $B_2$ appears

The same telescoping trick, one power up: since $(k+1)^3-k^3 = 3k^2+3k+1$, summing from $k=1$ to $n$ gives

$$
(n+1)^3 - 1 = 3\sum_{k=1}^{n} k^2 + 3\sum_{k=1}^{n} k + n
$$

Substituting the $p=1$ result and solving for $\sum k^2$,

$$
\sum_{k=1}^{n} k^2 = \frac{n(n+1)(2n+1)}{6} = \frac{n^3}{3} + \frac{n^2}{2} + \frac{n}{6}
$$

Compare this to the $p=2$ case of Faulhaber's formula:

$$
\frac{1}{3}\Big(B_0 n^3 + 3B_1 n^2 + 3B_2 n\Big) = \frac{n^3}{3} + \frac{n^2}{2} + B_2\, n
$$

Matching coefficients of $n$, we read off $B_2 = \frac{1}{6}$ — the very same constant that will reappear in Part IV, where it is exactly what makes $\overset{R}{\sum} n = -\frac{1}{12}$. Bernoulli numbers are not exotic machinery invented just to make divergent sums behave; they already sit quietly inside the most ordinary sums of powers, and Part IV simply asks what happens when the same expansion is pushed to a case where the sum itself no longer converges.
