# Ramanujan Summation


In Part III we saw that Cesàro summation and Abel summation, while both legitimate ways of assigning
a finite value to certain divergent series, are not enough to make sense of $C = 1+2+3+\ldots$. Neither
the running averages of the partial sums, nor the limit as $x \to 1^-$ of the power series
$\sum_{n=1}^{\infty} n x^n$, converge to a finite number. We need a strictly more powerful notion.

This is where Ramanujan summation comes in. It is the method Ramanujan himself used, in a 1913 letter to
G. H. Hardy, to write down statements like $1+2+3+\ldots = -\frac{1}{12}$. Unlike Cesàro and Abel
summation, it is *not* a limit of any naturally associated sequence. It is instead a specific finite
quantity extracted algebraically from an asymptotic expansion. This makes it more powerful, but also
more delicate: the resulting "sum" does not obey all the same algebraic rules as an ordinary sum.

### The Euler-Maclaurin formula

The starting point is the Euler-Maclaurin formula, which relates a sum of a smooth function $f$ over
integers to an integral of $f$, plus a boundary term, plus a correction term:

$$
\sum_{k=1}^{n} f(k) = \int_1^n f(x)\,dx + \frac{f(1)+f(n)}{2} + \sum_{k=1}^{p} \frac{B_{2k}}{(2k)!}\Big(f^{(2k-1)}(n)-f^{(2k-1)}(1)\Big) + R_p(n)
$$

Here $B_{2k}$ are the Bernoulli numbers, and $R_p(n)$ is a remainder term that, for well-behaved $f$,
stays bounded as $n \to \infty$.

The parameter $p$ is simply how many terms of this expansion you choose to keep before stopping — a free
cutoff, not something determined by $f$. This is not a convergent series in $p$ in general: the Bernoulli
terms typically grow rather than shrink as $p\to\infty$, so the formula is an *asymptotic* expansion, not
a Taylor-style series you'd want to sum to infinity. The correct way to use it is to fix $p$ at some small
value, and only then let $n\to\infty$; the remainder $R_p(n)$ is guaranteed to stay under control for that
fixed $p$. For a polynomial $f$, this is moot: derivatives of order $\geq \deg f$ vanish identically, so
the expansion terminates on its own and any $p$ large enough to include all nonzero terms gives an exact
(not just asymptotic) formula.

The key observation is that the only term on the right that depends on $n$ in an unbounded way is the
integral $\int_1^n f(x)\,dx$. Everything else either stays bounded, or depends only on $f(1)$ and its
derivatives at $1$. This suggests a way to separate the "divergent part" of the sum (the integral) from
a leftover "constant part" that captures everything else.</cell id="6d9c0804">


### The Ramanujan constant

Apply the Euler-Maclaurin formula with the integral anchored at $0$ instead of $1$:

$$
\sum_{k=1}^{n} f(k) = \int_0^n f(x)\,dx + \frac{f(n)+f(0)}{2} + \sum_{m=1}^{p} \frac{B_{2m}}{(2m)!}\Big(f^{(2m-1)}(n)-f^{(2m-1)}(0)\Big) + R_p(n)
$$

As before, every term depending on $n$ either diverges (the integral) or is the "regular", $n$-dependent
part of an asymptotic expansion. What's left over — the part that depends only on $f$ and its behavior
at $0$ — is what Ramanujan himself (and Hardy, writing it up rigorously afterward) called the *constant
of the series*:

**Definition.** For a suitably smooth function $f$, the *Ramanujan constant* is

$$
R(f) := -\frac{1}{2}f(0) - \sum_{m=1}^{\infty} \frac{B_{2m}}{(2m)!}\,f^{(2m-1)}(0)
$$

and the *Ramanujan sum* of $\sum_{n \geq 1} f(n)$ is written $\overset{R}{\sum_{n=1}^{\infty}} f(n) := R(f)$.

As with the earlier, more modern formulation, this is not a limit of partial sums — it is a specific
constant extracted algebraically from an asymptotic expansion, defined whenever the series of Bernoulli
terms above makes sense (e.g. terminates, as it does for polynomial $f$, or converges).

A word of warning: this is *not* the same convention as some other rigorous treatments of "the Ramanujan
summation" (for instance Candelpergher's, which anchors the Euler-Maclaurin formula at $x=1$ using the
periodized Bernoulli polynomial $B_1$ instead of derivatives at $0$). The two conventions can disagree —
we'll see exactly how much they disagree by, and why, once we work the example below.</cell id="be5d208d">


### Example: $f(x) = x$

Let's apply the definition to $f(x) = x$, so that $\overset{R}{\sum} f(n) = \overset{R}{\sum} n$ is a
candidate meaning for $C = 1+2+3+\ldots$

We have $f(0) = 0$, and $f'(x) = 1$ is constant, so every derivative of order $2$ or higher vanishes:
$f^{(2m-1)}(0) = 0$ for all $m \geq 2$. Only the $m=1$ term of the sum survives:

$$
R(f) = -\frac{1}{2}(0) - \frac{B_2}{2!}\,f'(0) = -\frac{B_2}{2}
$$

Using $B_2 = \frac{1}{6}$,

$$
R(f) = -\frac{1}{2}\cdot\frac{1}{6} = -\frac{1}{12}
$$

So, in the Ramanujan summation sense,

$$
\overset{R}{\sum_{n=1}^{\infty}} n = -\frac{1}{12}
$$

This is the precise, rigorous statement that sits behind the "paradox" from Part I. It is not saying
that the ordinary infinite sum $1+2+3+\ldots$ converges to $-\frac{1}{12}$ — it provably diverges, as we
showed at the very start. It is saying that $-\frac{1}{12}$ is the value of a specific, well-defined
regularization procedure applied to the sequence of positive integers.

#### Why the other convention gives a different number

It's worth being upfront about something most popular accounts gloss over: the more modern, rigorous
formulation of $C(f)$ from the previous section — anchored at $x=1$ rather than $x=0$ — does *not* give
$-\frac{1}{12}$ for this same series. Using that convention, one can show (via the family
$f_s(x) = x^{-s}$ and analytic continuation in $s$) that

$$
\overset{R}{\sum_{n=1}^{\infty}} n^{-s} = \zeta(s) - \frac{1}{s-1}, \qquad s \neq 1
$$

Setting $s=-1$ gives $C(f) = \zeta(-1) - \frac{1}{-1-1} = -\frac{1}{12} + \frac{1}{2} = \frac{5}{12}$ —
different from $R(f) = -\frac{1}{12}$ by exactly $\frac{1}{2}$.

This is not a contradiction, and it isn't a mistake in either formula — it is a real, well-documented
ambiguity that Hardy himself pointed out in Ramanujan's original notebooks: "the constant of a series"
depends on *where* you anchor the Euler-Maclaurin expansion, and different choices shift the answer by
a boundary term. $-\frac{1}{12}$ is the value from Ramanujan's own convention (anchored at $0$), and it
is also the value of $\zeta(-1)$ directly. $\frac{5}{12}$ is what you get from the alternative, arguably
more modern and systematic convention anchored at $1$. Both are legitimate regularizations; they simply
answer slightly different questions. The famous claim "$1+2+3+\ldots=-\frac{1}{12}$" is really a claim
about Ramanujan's original convention (equivalently, about $\zeta(-1)$) — not a universal, convention-free
fact about divergent series.</cell id="cadc743e">


### A word of caution

Ramanujan summation is powerful, but it does not behave like ordinary summation in every respect. In
particular, it is **not** shift-invariant: if $a_n$ is a sequence, the Ramanujan sum of $a_{n+1}$ is not
simply the Ramanujan sum of $a_n$ minus $a_1$, the way it would be for an ordinary convergent series.
This is precisely the property that the fallacious "summation trick" proofs in Part I implicitly (and
incorrectly) assumed held for all divergent series. Ramanujan summation comes with an explicit correction
term whenever you shift indices, and keeping careful track of that correction term is what separates the
false proofs from the true one.

This is also why the result is used in physics, in the context of zeta function regularization and the
Casimir effect: physicists are not claiming that infinitely many photon modes "really" sum to a negative
fraction. They are using a regularization procedure — this one, or an equivalent one — that reproduces
the correct, experimentally verified finite answer once genuinely divergent quantities are subtracted
off in a consistent way.
